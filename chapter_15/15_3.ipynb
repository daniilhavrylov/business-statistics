```python
import pandas as pd
import statsmodels.api as sm
```

y = natural gas consumption

x1 = average hourly temperature

x2 = the chill index

```python
df = pd.DataFrame({
    "y": [12.4, 11.7, 12.4, 10.8, 9.4, 9.5, 8.0, 7.5],
    "x1": [28.0, 28.0, 32.5, 39.0, 45.9, 57.8, 58.1, 62.5],
    "x2": [18, 14, 24, 22, 8, 16, 1, 0]
})
df
```

```python
X = df[["x1", "x2"]]
X = sm.add_constant(X)
y = df["y"]

model = sm.OLS(y, X).fit()
model.summary()
```

b1 = -0.0900

b2 = 0.0825

```python
model.model.exog_names
```

```python
htemperature = 40
chill_index = 10

prediction = model.get_prediction([1, htemperature, chill_index])
prediction.summary_frame(alpha=0.05)
```
