```python
import pandas as pd
import statsmodels.api as sm
```

y = sales price of a house (in thousands of dollars)

x1 = home size (in hundreds of square feet)

x2 = rating (an overall “niceness rating” for the house expressed on a scale from 1 [worst] to 10 [best], and provided by the real estate
agency)

```python
df = pd.DataFrame({
    "y":  [180, 98.1, 173.1, 136.5, 141, 165.9, 193.5, 127.8, 163.5, 172.5],
    "x1": [23, 11, 20, 17, 15, 21, 24, 13, 19, 25],
    "x2": [5,  2,  9,  3,  8,  4,  7,  6,  7,  2]
})
df
```

```python
X = df[["x1", "x2"]]
y = df["y"]

X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
model.summary()
```

b1 = 5.6128

b2 = 3.8344

```python
square_feet = 20
rating = 8

prediction = model.get_prediction([1, square_feet, rating])
prediction.summary_frame(alpha=0.05)
```
